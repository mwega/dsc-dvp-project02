```python
#Converting all column names to lowercase
df.columns = df.columns.str.lower()
df.head(5)
```

```python
# Creating new features for maize and sorghum yield per person 
df['s_yield_per_person'] = df['s_prod_tot'] / df['pop']
df['m_yield_per_person'] = df['m_prod_tot'] / df['pop']
df.head()
```

The additional maize and sorghum yield per person columns (s_yield_per_person and m_yield_per_person) measure how much sorghum and maize yield is available per person in each sub-county. This will help us understand whether the production is sufficient for the population in each area.

```python
# Checking for outliers using IQR method
def remove_outliers(df, s_yield_ha):
    Q1 = df[s_yield_ha].quantile(0.25)
    Q3 = df[s_yield_ha].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a boxplot for population
plt.figure(figsize=(8, 5))
sns.boxplot(x=df["pop"])
plt.title("Box Plot of Population")
plt.show()

# Boxplots for crop yield
plt.figure(figsize=(8, 5))
sns.boxplot(data=df[['s_yield_ha', 'm_yield_ha']])
plt.title("Box Plot of Sorghum & Maize Yield per Hectare")
plt.show()
```
